```python
import os
from dotenv import load_dotenv

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import JinaEmbeddings
```

```python
groq_key = os.getenv("GROQ_API_KEY")
jina_key = os.getenv("JINA_API_KEY")

print("env loaded successfully")
```

```python
DATA_FILE_PATH = os.path.join("data" , "data.pdf")
```

```python
loader = PyPDFLoader(DATA_FILE_PATH)
documents = loader.load()

print(documents)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)
print(chunks)
len(chunks)
```

```python
print(chunks[0])
```

```python
print(chunks[224].page_content)
```

```python
from langchain_community.embeddings import JinaEmbeddings

embeddings_model = JinaEmbeddings(model_name="jina-embeddings-v2-base-en")

print("embeddings model loaded successfully")
```

```python
from langchain_community.vectorstores import FAISS
vector_store = FAISS.from_documents(chunks , embeddings_model)
print("chunks are stored" , vector_store.index.ntotal , "in the vector store")
```

```python
test_query = "What is the fees for B.Tech?"
top_matches = vector_store.similarity_search(test_query , k=3)
print(f"Query: {test_query}")
for i, match in enumerate(top_matches,start=1):
    print(f"Match {i}: {match.page_content}...") 
    print()
```

```python
retriever = vector_store.as_retriever(search_kwargs={"k":3})

def Uniassist_retriever(question:str)-> str:
    """Search the document chunks for the answer to the question and return the matching chunks."""
    matching_chunks = retriever.invoke(question)
    return "/n/n".join([chunk.page_content for chunk in matching_chunks])
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.2,)
llm.model_name
```

```python
test_response = llm.invoke("Hey is learning rag hard?")
test_response.content
```

```python
from langchain.agents import create_agent
```

```python
UniAssist = create_agent(
    model=llm,
    tools=[Uniassist_retriever],
    system_prompt= """You are a helpful AI assistant which helps students during their admission process.
    Here is the context of the document you have to answer the questions based on it. 
    If you don't know the answer, just say that you don't know, don't try to make up an answer."""
    )

print("Ready to answer")
```

```python
def ask_UniAssist(question: str) -> str:
    """Send a question to the RAG agent and print a nicely formatted answer."""
    print("=" * 60)
    print("QUESTION:", question)
    print("-" * 60)

    response = ask_UniAssist.invoke({"messages": [{"role": "user", "content": question}]})
    answer = response["messages"][-1].content

    print("ANSWER:", answer)
    print("=" * 60)
    print()
    return answer
```

```python
response = UniAssist.invoke(
    {
        "messages":[
            {
                "role":"user",
                "content": "tell me which org you work for"
            }
        ]
    }
)
```
